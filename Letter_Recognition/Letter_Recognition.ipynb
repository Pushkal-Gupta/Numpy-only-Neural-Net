### Thus we can have an MLP of Architecture 16 -> 128 -> 64 -> 32 -> 26

#### Initializing Parameters

```python
# Using He Init
def init_params():
    W1 = np.random.randn(128,16) * np.sqrt(2/16)
    b1 = np.zeros((128,1))
    W2 = np.random.randn(64,128) * np.sqrt(2/128)
    b2 = np.zeros((64,1))
    W3 = np.random.randn(32,64) * np.sqrt(2/64)
    b3 = np.zeros((32,1))
    W4 = np.random.randn(26,32) * np.sqrt(2/32)
    b4 = np.zeros((26,1))
    
    wArr = [W1,W2,W3,W4]
    bArr = [b1,b2,b3,b4]
    
    return wArr,bArr
```

#### Utility Functions

```python
def ReLU(Z):
    return np.maximum(Z,0)
    
def derivative_ReLU(Z):
    return Z>0

def softmax(Z):
    expZ = np.exp(Z-np.max(Z,axis = 0, keepdims = True))
    return expZ/np.sum(expZ,axis = 0 ,keepdims = True)
    
def one_hot(Y):
    one_hot_Y = np.zeros((26,Y.size))
    
    for index,value in enumerate(Y):
        one_hot_Y[ord(value)-ord('A')][index] = 1
        
    return one_hot_Y
```

### Main Functions

```python
def forward_pass(X,wArr,bArr):
    
    #Number of Layers
    n = len(wArr)
    
    activationVal = X
    zArr = []
    aArr = []
    for i in range(n):
        
        zArr.append(wArr[i].dot(activationVal) + bArr[i])
        if i!=n-1:
            activationVal = ReLU(zArr[-1])
            aArr.append(activationVal)
        else:
            activationVal = softmax(zArr[-1])
            aArr.append(activationVal)
            
#     for index,arr in enumerate(aArr):
#         print(index," : ",len(arr),len(arr[0]))
        
    return zArr,aArr
```

```python
def backward_pass(X,Y,zArr,aArr,wArr):
    
    m = Y.size
    
    #Number of Layers
    n = len(wArr)
    
    dZArr,dWArr,dbArr = [],[],[]
    
    one_hot_Y = one_hot(Y)
    
    for i in range(n):
        
        if i==0:
            dZArr.append(aArr[-1] - one_hot_Y)
        else:
            dZArr.append(wArr[n-i].T.dot(dZArr[-1]) * derivative_ReLU(zArr[n-i-1]))
        
        if i==n-1:
            dWArr.append(1/m * dZArr[-1].dot(X.T))
        else:
            dWArr.append(1/m * dZArr[-1].dot(aArr[n-i-2].T))
            
        dbArr.append(1/m * np.sum(dZArr[-1], axis = 1, keepdims = True))
            
    return dWArr[::-1],dbArr[::-1]
```

```python
def update_params(alpha,dWArr,dbArr,wArr,bArr):
    
    #Number of Layers
    n = len(wArr)
    
    for i in range(n):
#         print("Debug w : ",len(wArr[i]),"x",len(wArr[i][0]),"Debug dw : ",len(dWArr[i]),"x",len(dWArr[i][0]))
        wArr[i] -= alpha * dWArr[i]
        bArr[i] -= alpha * dbArr[i]
        
    return wArr,bArr
```

### Driver Function

```python
def get_prediction(A):
    return np.argmax(A,0)

def get_accuracy(prediction,Y):
    #The Y values we get are letters
    digitY = [ord(char)-ord('A') for char in Y]
    
    return np.sum(prediction==digitY)/len(digitY)

def sgd(X,Y,alpha,epochs):
    
    wArr,bArr = init_params()
    batch_size = 100
        
    for epoch_cnt in range(epochs):
        
        shuffled_indices = np.random.permutation(Y.size)
        X_shuffled = X[:, shuffled_indices]
        Y_shuffled = Y[shuffled_indices]
    
        for start in range(0,Y.size,batch_size):

            end = start + batch_size

            X_batch = X_shuffled[:,start:end]
            Y_batch = Y_shuffled[start:end]

            #Perform Forward Pass
            zArr, aArr = forward_pass(X_batch,wArr,bArr)

            #Perform Backward Pass
            dWArr,dbArr = backward_pass(X_batch,Y_batch,zArr,aArr,wArr)

            #Update Parameters
            wArr,bArr = update_params(alpha,dWArr,dbArr,wArr,bArr)
        
        if epoch_cnt%10==0:
            
            print("Epoch : ",epoch_cnt)
            prediction = get_prediction(aArr[-1])
            print(get_accuracy(prediction,Y_batch))
            
    return wArr,bArr
```

```python
wArr,bArr = sgd(X_train,Y_train,0.02,200)
```

### 99% ACCURACY !!!!

```python
def make_prediction(X,wArr,bArr):
    
    #Perform Forward Pass
    _,aArr = forward_pass(X,wArr,bArr)
    A = aArr[-1]
    
    return get_prediction(A)
```

```python
predictions = make_prediction(X_test,wArr,bArr)
print(predictions.size,predictions,Y_test)
get_accuracy(predictions,Y_test)
```
